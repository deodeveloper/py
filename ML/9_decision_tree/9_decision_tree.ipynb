```python
import pandas as pd
```

```python
df = pd.read_csv("salaries.csv")
df.head()
```

```python
inputs = df.drop('salary_more_then_100k',axis='columns')
```

```python
target = df['salary_more_then_100k']
```

```python
from sklearn.preprocessing import LabelEncoder
le_company = LabelEncoder()
le_job = LabelEncoder()
le_degree = LabelEncoder()
```

```python
inputs['company_n'] = le_company.fit_transform(inputs['company'])
inputs['job_n'] = le_job.fit_transform(inputs['job'])
inputs['degree_n'] = le_degree.fit_transform(inputs['degree'])
```

```python
inputs
```

```python
inputs_n = inputs.drop(['company','job','degree'],axis='columns')
```

```python
inputs_n
```

```python
target
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(inputs_n, target)
```

```python
model.score(inputs_n,target)
```

**Is salary of Google, Computer Engineer, Bachelors degree > 100 k ?**

```python
model.predict([[2,1,0]])
```

**Is salary of Google, Computer Engineer, Masters degree > 100 k ?**

```python
model.predict([[2,1,1]])
```

```python
# load titanic data
import pandas as pd
dftitanic = pd.read_csv("./Exercise/titanic.csv")
dftitanic
```

```python
dftitanic = dftitanic.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns')
dftitanic
```

```python
dftitanic_inputs = dftitanic.drop('Survived',axis='columns')
dftitanic_target = dftitanic['Survived']
```

```python
dftitanic_inputs
```

```python
from sklearn.preprocessing import LabelEncoder
le_sex = LabelEncoder()
# encode sex column with le_sex
dftitanic_inputs["sex_n"] = le_sex.fit_transform(dftitanic["Sex"])
```

```python
dftitanic_inputs = dftitanic_inputs.drop('Sex', axis='columns', errors='ignore')
dftitanic_inputs
```

```python
# fill missing age values with mean
dftitanic_inputs.Age = dftitanic_inputs.Age.fillna(dftitanic_inputs.Age.mean())
dftitanic_inputs
```

```python
# generate x_train, x_test, y_train, y_test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(dftitanic_inputs, dftitanic_target, test_size=0.2)
```

```python
# create decision tree model
from sklearn import tree
model = tree.DecisionTreeClassifier()
# train model
model.fit(x_train, y_train)
# score model
model.score(x_test, y_test)
```
